# Introducción a sistemas expertos y Experta

Un **sistema experto** es un programa capaz de relacionar un conjunto de hechos con un conjunto de reglas aplicables y ejecutar acciones basadas en las reglas que coincidan con los hechos. En otras palabras, el conocimiento (reglas) y los datos (hechos) están separados, y un motor de inferencia (como KnowledgeEngine en Experta) aplica las reglas a los hechos para deducir conclusiones. 

**Experta** es una biblioteca de Python diseñada para construir sistemas expertos inspirada en CLIPS. Utilizaremos Experta para definir hechos (datos) y reglas (conocimiento) en un ejemplo didáctico.

### Definición de hechos

```python
# !pip install experta

import collections

if not hasattr(collections, 'Mapping'):
    import collections.abc
    collections.Mapping = collections.abc.Mapping

from experta import *

class Usuario(Fact):
    """Información sobre el estilo de vida del usuario."""
    pass
```

### Definición del motor de reglas (KnowledgeEngine)

Para crear el motor de inferencia definimos una clase que hereda de KnowledgeEngine. Cada método de esta clase decorado con @Rule(...) representa una regla. En el decorador @Rule se especifica la condición (LHS) mediante patrones de hechos, y el cuerpo del método es la acción (RHS) que se ejecuta cuando la regla dispara. 

```python
class SistemaSalud(KnowledgeEngine):
    
    # REGLA 1: Estilo de vida Sedentario
    @Rule(Usuario(horas_ejercicio=L(0)) | Usuario(trabajo="oficina"))
    def regla_sedentario(self):
        self.declare(Fact(perfil="sedentario"))
        print("-> Diagnóstico: Perfil detectado como Sedentario.")

    # REGLA 2: Riesgo por falta de sueño
    @Rule(Usuario(horas_sueno=L(range(0, 6))))
    def regla_insomnio(self):
        self.declare(Fact(riesgo_estres="alto"))
        print("-> Diagnóstico: Riesgo de estrés elevado por falta de sueño.")

    # REGLA 3: Recomendación de actividad
    @Rule(Fact(perfil="sedentario"),
          Fact(riesgo_estres="alto"))
    def recomendacion_suave(self):
        print("\n*** RECOMENDACIÓN FINAL ***")
        print("Tu nivel de actividad es bajo y tu estrés alto.")
        print("Se recomienda empezar con: Caminatas de 20 min y Yoga.")

    # REGLA 4: Atleta con poco sueño
    @Rule(Usuario(horas_ejercicio=P(lambda x: x > 5)),
          Fact(riesgo_estres="alto"))
    def advertencia_sobreentrenamiento(self):
        print("\n*** ADVERTENCIA ***")
        print("Entrenas mucho pero duermes poco. ¡Peligro de lesión!")
```

```python
# SIMULACIÓN  
def ejecutar_test():
    engine = SistemaSalud()
    engine.reset()
    
    print("Introduciendo hechos: Trabaja en oficina, 0 horas ejercicio, duerme 5 horas...")
    
    # Declaramos los hechos iniciales
    engine.declare(Usuario(trabajo="oficina", horas_ejercicio=0, horas_sueno=5))
    
    # Ejecutamos el motor
    engine.run()

if __name__ == "__main__":
    ejecutar_test()
```
