## Import Library







```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Import data from the dataset






```
# upload files from your local machine
from google.colab import files
#files.upload()
dataset = pd.read_excel('/content/1-Ensayo_Motores_rpm_par.xlsx')
X = dataset.iloc[:, [1,2]].values
y = dataset.iloc[:, 3].values
```

```
#Split data into training/testing
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

## KERNEL NO LINEAL.

## Creating and fitting the model

```
# Create a SVC object
from sklearn.svm import SVC
classifierSVC = SVC(kernel = "rbf", probability=True, random_state = 0)
classifierSVC.fit(X_train_sc, y_train)
#parametro gamma = anchura de la campana de rbf
```

**Parametro gamma**

- **Gamma alto**: Cada punto influye solo en su entorno inmediato, generando **fronteras de decisión (plano de corte)** muy complejas y riesgo de sobreajuste.
  - Hace que la función del kernel decaiga rápidamente conforme aumenta la distancia entre puntos.
  - El **kernel** es una función que mide la similitud entre dos puntos en el espacio de características y permite, de forma implícita, proyectar esos datos a un espacio de mayor dimensión sin tener que realizar esa transformación de manera explícita.
- **Gamma bajo**: La influencia de cada punto se extiende más, produciendo fronteras de decisión suaves y mayor capacidad de generalización.


image.png

```
# Obtener las probabilidades de la clase 1 (positiva)
y_proba = classifierSVC.predict_proba(X_train_sc)[:, 1]

# Graficar un histograma de probabilidades
plt.hist(y_proba, bins=20, alpha=0.7, color='blue', edgecolor='black')
plt.xlabel("Probabilidad de fallo del motor")
plt.ylabel("Número de motores")
plt.title("Distribución de Probabilidades de Fallo del Motor")
plt.show()
```

El gráfico muestra una distribución bimodal, donde el modelo predice con alta confianza (probabilidades cercanas a 0 o 1) si un motor fallará o no. Parece haber una correlación entre RPM altas (~150,000) y fallos del motor, mientras que motores con RPM bajas son clasificados como no fallidos. Esto sugiere que el modelo ha captado un patrón claro

Internamente, `cross_val_score` realiza los siguientes pasos para cada partición:

1. **Divide los datos**: La función divide el conjunto de datos en un número especificado de "folds" o particiones. Por defecto, se utiliza la validación cruzada K-Fold, que divide los datos en K partes iguales.

2. **Entrena el modelo**: Para cada fold, `cross_val_score` entrena el modelo utilizando todos los datos excepto los del fold actual, que se reservan para la prueba.

3. **Evalúa el modelo**: Una vez entrenado el modelo, se utiliza el fold reservado para la prueba para evaluar su rendimiento. Aquí es donde `cross_val_score` efectivamente ejecuta el método `predict` del modelo para hacer predicciones sobre los datos de prueba.

4. **Calcula la puntuación**: Después de hacer las predicciones, `cross_val_score` calcula una puntuación de rendimiento, como la precisión, utilizando las predicciones y las verdaderas etiquetas de los datos de prueba.

5. **Repite y promedia**: Este proceso se repite para cada fold, y luego `cross_val_score` devuelve la media de todas las puntuaciones obtenidas para los diferentes folds.

```
from sklearn.model_selection import cross_val_score
```

```
scores = cross_val_score(LogisticRegression(random_state = 0), X, y, scoring="accuracy", cv=10)
```

**cross_val_score(...)**

evalúa un modelo mediante validación cruzada y devuelve una lista de puntuaciones (una por cada iteración).

**cv=10**

- cv=10 significa que se usa validación cruzada de 10 folds (10-fold cross-validation).
- Proceso:
  1. Se divide X y y en 10 partes (folds).
  2. En cada iteración, se toma un fold como conjunto de prueba y los 9 restantes como entrenamiento.
  3. Se repite 10 veces (cada fold es usado una vez para prueba).
  4. Se devuelve una lista con 10 valores de accuracy (uno por cada iteración).
