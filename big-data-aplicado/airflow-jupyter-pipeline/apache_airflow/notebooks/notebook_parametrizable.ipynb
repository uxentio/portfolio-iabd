# Notebook parametrizable

Este notebook recibe parametros desde Airflow mediante Papermill y realiza una tarea sencilla con ellos.

```python
# Celda de parametros - Papermill inyecta los valores aqui
nombre = "Mundo"
repeticiones = 3
```

```python
# Mostramos los parametros recibidos
print(f"Nombre recibido: {nombre}")
print(f"Repeticiones: {repeticiones}")
```

```python
# Ejecutamos la tarea con los parametros
for i in range(repeticiones):
    print(f"Iteracion {i+1}: Hola, {nombre}!")
```

```python
# Resultado final
resultado = f"Se ejecutaron {repeticiones} iteraciones para {nombre}"
print(resultado)
```
